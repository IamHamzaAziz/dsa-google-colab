# Without Collision Handling

```python
class HashTable:
  def __init__(self, size):
    self.size = size
    self.arr = [None for i in range(size)]

  # __ so that this function can only be used within this class
  def __get_hash(self, key):
    h = 0
    for char in key:
      h += ord(char)  # ord gives the ASCII number of provided element

    return h % self.size

  def add(self, key, value):
    hashed_key = self.__get_hash(key)
    self.arr[hashed_key] = value

  def get(self, key):
    hashed_key = self.__get_hash(key)
    return self.arr[hashed_key]

  # for setting data like obj[key] = value
  def __setitem__(self, key, value):
    hashed_key = self.__get_hash(key)
    self.arr[hashed_key] = value

  # for getting data like obj[key]
  def __getitem__(self, key):
    hashed_key = self.__get_hash(key)
    return self.arr[hashed_key]

  def __delitem__(self, key):
    hashed_key = self.__get_hash(key)
    self.arr[hashed_key] = None
```

```python
ht = HashTable(20)
ht.add('March 1', 80)
ht.add('March 13', 52)
ht.add('June 20', 36)
ht.add('August 7', 20)
ht.add('October 17', 93)

print('June 20:', ht.get('June 20'))

ht['October 30'] = 45
print('October 30:', ht['October 30'])

del ht['August 7']
print('August 7:', ht['August 7'])
```

# With Collision handling

```python
class HashTable:
  def __init__(self, size):
    self.size = size
    self.arr = [[] for i in range(size)]

  def __hash(self, key):
    h = 0
    for char in key:
      h += ord(char)
    return h % self.size

  def set(self, key, element):
    hashed_key = self.__hash(key)

    if len(self.arr[hashed_key]) == 0:
      self.arr[hashed_key].append((key, element))
    else:
      found = False
      for i in range(len(self.arr[hashed_key])):
        if self.arr[hashed_key][i][0] == key:
          self.arr[hashed_key][i] = (key, element)
          found = True
          break

      if not found:
        self.arr[hashed_key].append((key, element))

  def get(self, key):
    hashed_key = self.__hash(key)

    if len(self.arr[hashed_key]) == 0:
      return self.arr[hashed_key][0][1]
    else:
      for idx, element in enumerate(self.arr[hashed_key]):
        if element[0] == key:
          return element[1]

      raise Exception('Invalid element search')

  def __getitem__(self, key):
    hashed_key = self.__hash(key)

    if len(self.arr[hashed_key]) == 0:
      return self.arr[hashed_key][0][1]
    else:
      for idx, element in enumerate(self.arr[hashed_key]):
        if element[0] == key:
          return element[1]

      raise Exception('Invalid element search')

  def __setitem__(self, key, element):
    hashed_key = self.__hash(key)

    if len(self.arr[hashed_key]) == 0:
      self.arr[hashed_key].append((key, element))
    else:
      found = False
      for i in range(len(self.arr[hashed_key])):
        if self.arr[hashed_key][i][0] == key:
          self.arr[hashed_key][i] = (key, element)
          found = True
          break

      if not found:
        self.arr[hashed_key].append((key, element))

  def __delitem__(self, key):
    hashed_key = self.__hash(key)

    if len(self.arr[hashed_key]) == 1:
      self.arr[hashed_key] = []
    else:
      for i in range(len(self.arr[hashed_key])):
        if self.arr[hashed_key][i][0] == key:
          del self.arr[hashed_key][i]
          break
```

```python
ht = HashTable(10)

ht.set("march 6", 120)
ht.set("march 17", 20)
ht.set("march 9", 150)
ht.set("march 6", 160)
```
