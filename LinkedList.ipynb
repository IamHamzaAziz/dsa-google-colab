```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class LinkedList:
  def __init__(self):
    self.head = None

  def insert_at_beginning(self, data):
    node = Node(data)
    node.next = self.head
    self.head = node

  def insert_at_end(self, data):
    node = Node(data)

    if self.head is None:
      self.head = node
      return

    itr = self.head
    while itr.next:
      itr = itr.next

    itr.next = node

  def insert_at(self, index, data):
    if index < 0 and index >= self.get_length():
      raise Exception('Invalid Index')

    if index == 0:
      self.insert_at_beginning(data)
      return

    itr = self.head
    counter = 0
    while counter < index - 1:
      itr = itr.next
      counter += 1

    node = Node(data)
    node.next = itr.next
    itr.next = node

  def insert_values(self, data_list):
    for i in data_list:
      self.insert_at_end(i)

  def get_length(self):
    counter = 0
    if self.head is None:
      return counter

    itr = self.head
    while itr:
      itr = itr.next
      counter += 1

    return counter

  def remove_at(self, index):
    if self.head is None:
      raise Exception('Linked list is empty')

    if index < 0 and index >= self.get_length():
      raise Exception('Invalid Index')

    if index == 0:
      self.head = self.head.next

    itr = self.head
    counter = 0
    # while counter < index - 1:
    #   itr = itr.next
    #   counter += 1
    # itr.next = itr.next.next

    # or
    while itr:
      if counter == index - 1:
        itr.next = itr.next.next
        break

      counter += 1
      itr = itr.next


  def print_linked_list(self):
    if self.head is None:
      print("Linked list is empty")
      return

    itr = self.head

    while itr:
      print(itr.data, end=" ")
      itr = itr.next
```

```python
ll = LinkedList()

ll.insert_at_beginning(5)
ll.insert_at_beginning(89)
ll.insert_at_end(45)
ll.insert_at_beginning(19)

ll.insert_values([76, 87, 6])

print('Lenght of ll is', ll.get_length())
ll.print_linked_list()

print()

ll.remove_at(2)
ll.print_linked_list()

print()

ll.insert_at(2, 4)
ll.print_linked_list()

print()
```

## Number 1
Create two functions in above LinkedList class.

First will insert a node after a specific value (or first occurance of that value).

Second function will remove the node with a specific value (or first occurance of that value)

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class LinkedList:
  def __init__(self):
    self.head = None

  def insert_at_beginning(self, data):
    node = Node(data)
    if self.head is None:
      self.head = node
      return

    node.next = self.head
    self.head = node

  def print_values(self):
    if self.head is None:
      raise Exception('Linked List is empty')

    itr = self.head
    while itr:
      # Last node will have next equal to None
      # These conditions will make sure last data will not have suffix ->
      if itr.next:
        print(itr.data, end='->')
      else:
        print(itr.data)
      itr = itr.next

  def insert_after_value(self, value, data):
    if self.head is None:
      raise Exception('Linked List is empty')

    if self.head.data == value:
      node = Node(data)
      node.next = self.head.next
      self.head.next = node
      return

    itr = self.head
    while itr:
      if itr.data == value:
        node = Node(data)
        node.next = itr.next
        itr.next = node
        break

      itr = itr.next

  def remove_by_value(self, value):
    if self.head is None:
      raise Exception('Linked List is empty')

    if self.head.data == value:
      self.head = self.head.next
      return

    itr = self.head
    while itr:
      if itr.next.data == value:
        itr.next = itr.next.next
        return
      itr = itr.next

  def reverse(self):
    if self.head is None:
      raise Exception('Linked List is empty')

    prev = None
    current = self.head

    while current:
      next = current.next
      current.next = prev
      prev = current
      current = next

    self.head = prev

```

```python
ll = LinkedList()

ll.insert_at_beginning(4)
ll.insert_at_beginning(5)
ll.insert_at_beginning(6)
ll.insert_at_beginning(7)
ll.print_values()

ll.insert_after_value(5, 9)
ll.print_values()

ll.insert_after_value(7, 65)
ll.print_values()

ll.insert_after_value(4, 56)
ll.print_values()

ll.remove_by_value(6)
ll.print_values()

ll.remove_by_value(7)
ll.print_values()

ll.reverse()
ll.print_values()
```

## Double Linked List
Create a double or doubly linked list

```python
class Node:
  def __init__(self, data):
    self.next = None
    self.prev = None
    self.data = data

class DoublyLinkedList:
  def __init__(self):
    self.head = None

  def insert_at_beginning(self, data):
    node = Node(data)
    if self.head is None:
      self.head = node
      return

    node.next = self.head
    self.head.prev = node
    self.head = node

  def print_linked_list(self):
    if self.head is None:
      raise Exception('Linked List is empty')

    itr = self.head
    while itr:
      if itr.next:
        print(itr.data, end="<->")
      else:
        print(itr.data)
      itr = itr.next

  def reverse(self):
    if self.head is None:
      raise Exception('Linked List is empty')

    current = self.head
    prev = None
    while current:
      next = current.next
      current.next = prev
      current.prev = next
      prev = current
      current = next

    self.head = prev
```

```python
dll = DoublyLinkedList()

dll.insert_at_beginning(5)
dll.insert_at_beginning(6)
dll.insert_at_beginning(7)
dll.insert_at_beginning(1)

dll.print_linked_list()

print()

dll.reverse()
dll.print_linked_list()
```
