In Python we have a built in data structure list which kind of works in similar way as stack.

However it uses dynamic array which is not good when size of list is maxed and we want to insert any other element. For that it creates duplicate copy with same size.

However collections.deque in python uses doubly linked list which is better in this regard

```python
from collections import deque
```

```python
arr = deque()
```

```python
arr.append(1)
arr.append(2)
arr.append(3)
arr.append(7)

arr
```

```python
class Stack:
  def __init__(self):
    self.container = deque()

  def push(self, element):
    self.container.append(element)

  def pop(self):
    return self.container.pop()

  # Return last element
  def peek(self):
    return self.container[-1]

  def is_empty(self):
    return len(self.container) == 0

  def size(self):
    return len(self.container)
```

```python
class Stack:
  def __init__(self):
    self.container = deque()

  def push(self, element):
    self.container.append(element)

  def pop(self):
    return self.container.pop()

  # Return last element
  def peek(self):
    return self.container[-1]

  def is_empty(self):
    return len(self.container) == 0

  def size(self):
    return len(self.container)
```

Write a function in python that can reverse a string using stack data structure.

```python
def reverse_string(s):
  input_string = Stack()

  for char in s:
    input_string.push(char)

  reversed = ''

  while input_string.size() != 0:
    reversed += input_string.pop()

  return reversed
```

```python
print("hola reversed:", reverse_string('hola'))
```

Write a function in python that checks if paranthesis in the string are balanced or not. Possible parantheses are "{}',"()" or "[]".

```python
def is_match(char1, char2):
  match_dist = {
      "{": "}",
      "[": "]",
      "(": ")"
  }

  return match_dist[char1] == char2

def is_balanced(s):
  stack = Stack()

  for i in s:
    if i == '{' or i == '(' or i == '[':
      stack.push(i)

    if i == '}' or i == ')' or i == ']':
      if stack.size() == 0:
        return False
      if not is_match(stack.pop(), i):
        return False

  return stack.size()==0
```

```python
print(is_balanced("({a+b})"))
print(is_balanced("))((a+b}{"))
print(is_balanced("((a+b))"))
print(is_balanced("((a+g))"))
print(is_balanced("))"))
print(is_balanced("[a+b]*(x+2y)*{gg+kk}"))
```
